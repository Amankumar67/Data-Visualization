```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]

sales = [100, 150, 120, 180, 200]
profit = [20, 30, 25, 40, 50]
expenses = [80, 100, 95, 110, 120]
students = [40, 45, 50, 48, 55]
attendance = [85, 90, 78, 92, 88]
marks = [60, 70, 65, 80, 85]

plt.subplot(2, 3, 1)
plt.plot(x, sales)
plt.title("Sales")

plt.subplot(2, 3, 2)
plt.bar(x, profit)
plt.title("Profit")

plt.subplot(2, 3, 3)
plt.plot(x, expenses)
plt.title("Expenses")

plt.subplot(2, 3, 4)
plt.bar(x, students)
plt.title("Students")

plt.subplot(2, 3, 5)
plt.plot(x, attendance)
plt.title("Attendance")

plt.subplot(2, 3, 6)
plt.bar(x, marks)
plt.title("Marks")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]
sales = [100, 150, 130, 180, 200]
profit = [20, 30, 25, 40, 50]

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(days, sales, marker="o",markerfacecolor="red")
plt.title("Sales")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.grid()

plt.subplot(1, 2, 2)
plt.bar(days, profit,color="blue",edgecolor="black")
plt.title("Profit")
plt.xlabel("Days")
plt.ylabel("Profit")
plt.grid()

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]
temperature = [25, 28, 30, 27, 32]
sales = [100, 150, 130, 180, 200]
hours = [1, 2, 3, 4, 5]
marks = [40, 50, 65, 75, 90]
expenses = [40, 30, 20, 10]

plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.plot(days, temperature, marker="o")
plt.title("Temperature")
plt.xlabel("Days")
plt.ylabel("Temperature")
plt.grid()

plt.subplot(2, 2, 2)
plt.bar(days, sales)
plt.title("Sales")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.grid()

plt.subplot(2, 2, 3)
plt.scatter(hours, marks,linewidth=8,color="brown",linestyle="--")
plt.title("Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid()

plt.subplot(2, 2, 4)
plt.pie(expenses, labels=["Rent", "Food", "Travel", "Other"], autopct="%1.1f%%")
plt.title("Expenses")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

sales = [100, 120, 150, 140, 180, 200]
profit = [20, 25, 35, 30, 45, 50]
expenses = [80, 95, 115, 110, 135, 150]

plt.figure(figsize=(10, 10))

plt.subplot(3, 1, 1)
plt.plot(months, sales, marker="o")
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid()

plt.subplot(3, 1, 2)
plt.bar(months, profit)
plt.title("Monthly Profit")
plt.xlabel("Months")
plt.ylabel("Profit")
plt.grid()

plt.subplot(3, 1, 3)
plt.fill_between(months, expenses, alpha=0.5)
plt.title("Monthly Expenses")
plt.xlabel("Months")
plt.ylabel("Expenses")
plt.grid()

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]
sales = [100, 150, 130, 180, 200]

hours = [1, 2, 3, 4, 5]
marks = [40, 50, 65, 75, 90]

scores = [45, 50, 55, 60, 65, 70, 72, 75, 78, 80, 85, 90]

subjects = ["Python", "SQL", "Pandas", "NumPy"]
marks_subject = [90, 75, 85, 80]

plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.plot(days, sales, marker="o")
plt.title("Sales Line Plot")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.grid()

plt.subplot(2, 2, 2)
plt.scatter(hours, marks)
plt.title("Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid()

plt.subplot(2, 2, 3)
plt.hist(scores, bins=5, edgecolor="black")
plt.title("Score Distribution")
plt.xlabel("Scores")
plt.ylabel("Frequency")
plt.grid()

plt.subplot(2, 2, 4)
plt.bar(subjects, marks_subject)
plt.title("Subject Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.grid()

plt.tight_layout()
plt.show()
```
