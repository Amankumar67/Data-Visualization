#                                                    LINE PLOT

```python
import matplotlib.pyplot as plt

days = ["Monday", "Tuesday", "Wednesday", "Thursday",
        "Friday", "Saturday", "Sunday"]
sales = [120, 150, 140, 180, 200, 220, 250]
plt.plot(
    days,
    sales,
    color="orange",
 
)
plt.title("Weekly Sales Report")
plt.xlabel("Days")
plt.ylabel("Sales of the Week")

plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt 
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperature = [28, 30, 29, 32, 31, 33, 35]
plt.plot(days,temperature,marker = "*",label = "temperature",color="green",markerfacecolor="blue",markeredgecolor="red")
plt.xlabel("days",color="blue",fontsize=12)
plt.ylabel("Temperature (°C)")
plt.title("weekly temperature report")
plt.legend()
plt.xticks(color ="purple",rotation = 45,fontsize = 16)
plt.grid()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
sales = [1200, 1000, 1300, 1450, 1250, 1200, 1100]

plt.plot(days,sales,marker = "*",markersize = 12,color = "black")

plt.title("Weekly Sales",color = "blue",fontsize = 18 )
plt.xlabel("Days",color = "blue", fontsize = 16)
plt.ylabel("Sales",color = "blue",fontsize = 16)
plt.xticks(rotation = 45)
plt.grid()
plt.show()
```

```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
data = pd.DataFrame({
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'Jan', 'Feb', 'Mar', 'Apr'],
    'Sales': [100, 120, 110, 160, 80, 110, 130, 150],
    'Product': ['A', 'A', 'A', 'A', 'B', 'B', 'B', 'B']
})

sns.lineplot(
    data=data,
    x='Month',
    y='Sales',
    hue='Product',
    marker='o'
)

plt.title("Product Sales")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri']

product_a = [20, 30, 25, 40, 50]
product_b = [15, 25, 35, 30, 45]
product_c = [18, 20, 34, 26, 42]

plt.plot(days, product_a, marker='o', label='Product A',linestyle="--")
plt.plot(days, product_b, marker='o', label='Product B',linestyle=":")
plt.plot(days, product_c, marker="*", label='Product C',linestyle="dashdot")

plt.title("Product Sales Comparison")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt 
Days = ["Mon", "Tue", "Wed", "Thu", "Fri"]

Python = [20,30,25,40,50]
Java   = [15,25,30,35,40]
C   = [10,20,35,30,45]

plt.plot(Days, Python,label="python",marker = "*",linestyle = "dashdot")
plt.plot(Days, Java,label = "java",marker = ">",linestyle = "--")
plt.plot(Days, C,label = "C",marker = "o",linestyle = ":")
plt.title("Multiple Line Plot",color = "blue", fontsize = 18)
plt.xlabel("Days",color = "blue",fontsize = 16)
plt.ylabel("Scores",color = "blue", fontsize = 17)
plt.ylim(0,60)
plt.legend()
plt.show()
```

```python

import matplotlib.pyplot as plt 
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

python = [65, 70, 75, 80, 85, 90]
java = [60, 65, 68, 72, 76, 80]
cpp = [55, 62, 67, 70, 74, 78]
plt.plot(months,python,marker = "*",color = "purple",label= "python",linestyle="dashdot")
plt.plot(months,java,marker="o",color = "green",label = "java",linestyle = "dashed")
plt.plot(months,cpp,marker = ">",color = "brown",label = "c++",linestyle = ":")
plt.title("Multiples lines")
plt.legend()
plt.grid(color= "blue",alpha = 0.4)
plt.show()
```

```python
import matplotlib.pyplot as plt

students = ["Aman", "Rahul", "Priya", "Neha", "Rohit"]
python_marks = [65, 72, 80, 75, 90]
java_marks = [60, 68, 78, 70, 85]
plt.figure(figsize=(9, 5))
plt.plot(students, python_marks,
         marker="o",
         linestyle="-",
         linewidth=2,
         label="Python")
plt.plot(students, java_marks,
         marker=">",
         linestyle="--",
         linewidth=2,
         label="Java")
plt.title("Student Performance Comparison")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.ylim(50, 100)
plt.grid(True, linestyle=":")
plt.legend()
plt.show()
```
