```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5,6,7,8,9,2,3,5,6,7,8,9,1]
y = [10, 20, 15, 30, 25,23,45,24,35,45,67,64,65,12,34,21,26]

plt.step(x, y,color="blue",linewidth=2,marker="o",label="result")
plt.scatter(x,y,color="green",linewidth=4,marker=">",edgecolor="red",s=70,label="sales")
plt.colorbar()
plt.xlim(0,10)
plt.title("marks distrubution")
plt.xlabel("students")
plt.ylabel("marks")
plt.legend(loc="upper left")
plt.grid(axis="both",color="black",alpha = 0.7)
plt.show()
```

```python
import matplotlib.pyplot as plt
time = [1, 2, 3, 4, 5, 6]
temperature = [20, 22, 22, 25, 25, 28]
plt.step(time,temperature,
         where="pre",
        zorder=1,
        alpha = 0.6,
        color="red",
        label="sales",
        linewidth = 3,
        linestyle=":",
        marker=">",
        markeredgecolor="green",
        markerfacecolor="blue")
plt.title("sales record",fontsize=12,color="brown")
plt.xlabel("days",fontsize=14,color="brown")
plt.ylabel("score",fontsize=14, color="brown")
plt.xticks(rotation=45,fontsize=12,color="blue")
plt.yticks(rotation=45,fontsize=12,color="blue")
plt.legend(loc="upper left", fontsize= 10)
plt.show()
```

```python
import matplotlib.pyplot as plt 
days = [1, 2, 3, 4, 5, 6]

sales = [100, 120, 150, 140, 180, 200]
profit = [20, 25, 30, 28, 35, 40]
plt.step(days,sales,where="post",color="red",marker="o",label="sales")
plt.step(days,profit,where="pre",color="green",marker=">",label="profit")
plt.title("report")
plt.xlabel("days")
plt.ylabel("sales and profit")
plt.grid(True)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

days = [1, 2, 3, 4, 5]
sales = [50, 80, 70, 100, 120]

plt.subplot(2, 2, 1)
plt.step(days, sales, where="pre")
plt.title("Pre")

plt.subplot(2, 2, 2)
plt.step(days, sales, where="post")
plt.title("Post")

plt.subplot(2, 2, 3)
plt.step(days, sales, where="mid")
plt.title("Mid")

plt.subplot(2, 2, 4)
plt.plot(days, sales)
plt.title("Normal Line")

plt.tight_layout()
plt.show()
```
