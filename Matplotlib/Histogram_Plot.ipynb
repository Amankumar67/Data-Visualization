```python
import matplotlib.pyplot as plt

python = [55, 60, 65, 70, 72, 75, 78, 80, 82, 85,
          88, 90, 92, 94, 96, 98]

java = [45, 50, 55, 58, 62, 65, 68, 70, 73, 76,
        78, 80, 83, 85, 88, 90]

plt.hist(python, bins=8, alpha=0.6, label="Python", edgecolor="black")
plt.hist(java, bins=8, alpha=0.6, label="Java", edgecolor="black")

plt.title("Python vs Java Scores")
plt.xlabel("Scores")
plt.ylabel("Number of Students")
plt.legend()

plt.show()
```
