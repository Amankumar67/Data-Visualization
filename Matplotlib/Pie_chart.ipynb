# Pie Chart 

```python
# Pie chart ka use data ko parts/percentage mein represent karne ke liye hota hai.
```

```python
import matplotlib.pyplot as plt 
time = [4,3,2,1]
data = ["python","dsa","dbms","others"]
plt.pie(time, labels = data,
       radius=1.2)  # it is used for changing the radius of pie 

plt.show()
```

```python
# autopct = for showing percentage in chart 
import matplotlib.pyplot as plt 
time = [4,3,2,1]
data = ["python","dsa","dbms","others"]
plt.pie(time, labels = data,autopct="%1.1f%%")
plt.show()
```

```python
# startangle controls the starting rotation/angle of the pie chart.
    
import matplotlib.pyplot as plt 
time = [4,3,2,1]
data = ["python","dsa","dbms","others"]
plt.pie(time, labels = data,autopct="%1.1f%%",startangle = 90 )
plt.show()
```

```python
# explode = it is used for separate any different value from others 
import matplotlib.pyplot as plt 
time = [4,3,2,1]
data = ["python","dsa","dbms","others"]
explode = [0.1, 0, 0, 0]

plt.pie(
    time,
    labels=data,
    autopct="%1.1f%%",
    startangle=90,
    explode=explode
)

plt.show()
```

```python
# explode = it is used for separate any different value from others 
import matplotlib.pyplot as plt 
time = [4,3,2,1]
data = ["python","dsa","dbms","others"]
explode = [0, 0.1, 0, 0]

plt.pie(
    time,
    labels=data,
    autopct="%1.1f%%",
    startangle=90,
    explode=explode
)

plt.show()
```

```python
# shadow = it is used for showing the background effecting .

import matplotlib.pyplot as plt 
time = [4,3,2,1]
data = ["python","dsa","dbms","others"]
explode = [0, 0.1, 0, 0]

plt.pie(
    time,
    labels=data,
    autopct="%1.1f%%",
    startangle=90,
    explode=explode,
    shadow = True
)

plt.show()
```

```python
# colors = it is used for sharing different color for each one 

import matplotlib.pyplot as plt 
time = [4,3,2,1]
data = ["python","dsa","dbms","others"]
explode = [0, 0.1, 0, 0]
colors = ["red", "blue", "green", "yellow"]



plt.pie(
    time,
    labels=data,
    autopct="%1.1f%%",
    startangle=90,
    explode=explode,
    shadow = True,
    colors = colors
)

plt.show()
```

```python
# wedgeprops = is used for making borders around slice 


import matplotlib.pyplot as plt 
time = [4,3,2,1]
data = ["python","dsa","dbms","others"]
explode = [0, 0.1, 0, 0]
colors = ["red", "blue", "green", "yellow"]



plt.pie(
    time,
    labels=data,
    autopct="%1.1f%%",
    startangle=90,
    explode=explode,
    shadow = True,
    colors = colors,
    wedgeprops = {
    "edgecolor": "black",
    "linewidth": 2
}
)

plt.show()
```

```python
# textprops = it is used to customize label and percenatge text.


import matplotlib.pyplot as plt 
time = [4,3,2,1]
data = ["python","dsa","dbms","others"]
explode = [0, 0.1, 0, 0]
colors = ["red", "blue", "green", "yellow"]



plt.pie(
    time,
    labels=data,
    autopct="%1.1f%%",
    startangle=90,
    explode=explode,
    shadow = True,
    colors = colors,
    wedgeprops = {
    "edgecolor": "black",
    "linewidth": 2},
textprops = {
    "fontsize": 10,
    "fontweight": "bold",
    "color":"black"}


)

plt.show()
```

```python
    # pctdistance = it is used for changing the way of percentage = 0.5-center , 0.7-near to edge , 1.0-on the edge
   # labeldistance = used for changing the position of label = 0.5-near to label, 1.0-on edge , 1.2-far from edge 
    # counterclock = it is used for decide then , the slice  of  pie chart must be clock wise or anti-clock wise ,
 # counterclock = false- clockwise direction , counterclock = True-anticlock-wise direction 
import matplotlib.pyplot as plt 
time = [4,3,2,1]
data = ["python","dsa","dbms","others"]
explode = [0, 0.1, 0, 0]
colors = ["red", "blue", "green", "yellow"]



plt.pie(
    time,
    labels=data,
    autopct="%1.1f%%",
    startangle=90,
    explode=explode,
    shadow = True,
    colors = colors,
    wedgeprops = {
    "edgecolor": "black",
    "linewidth": 2},
textprops = {
    "fontsize": 10,
    "fontweight": "bold",
    "color":"black"},
    pctdistance=0.5,
    labeldistance = 1.2,
    counterclock = True


)

plt.show()
```

```python
 # rotatelevels = is used for rotating to labels 
# if true = label slide ke according 
# if false = normal horizontal 

# normailze = false  Values ko normalize nahi karta; agar total 1 se kam ho to pie ka part incomplete ho sakta hai.
import matplotlib.pyplot as plt 
time = [4,3,1,1]
data = ["python","dsa","dbms","others"]
explode = [0, 0.1, 0, 0]
colors = ["red", "blue", "green", "yellow"]



plt.pie(
    time,
    labels=data,
    autopct="%1.1f%%",
    startangle=90,
    explode=explode,
    shadow = True,
    colors = colors,
    wedgeprops = {
    "edgecolor": "black",
    "linewidth": 2},
textprops = {
    "fontsize": 10,
    "fontweight": "bold",
    "color":"black"},
    pctdistance=0.5,
    labeldistance = 1.2,
    counterclock = True,
    rotatelabels = False,
    normalize = True


)
plt.legend(data)  # it is used for showing the data values as labels 
plt.show()
```

```python
import matplotlib.pyplot as plt

time = [4, 3, 2, 1]
data = ["Python", "DSA", "DBMS", "Others"]

explode = [0, 0.1, 0, 0]
colors = ["red", "blue", "green", "yellow"]

plt.pie(
    time,
    labels=data,
    autopct="%1.1f%%",
    startangle=90,
    explode=explode,
    shadow=True,
    colors=colors,
    wedgeprops={
        "edgecolor": "black",
        "linewidth": 2
    },
    textprops={
        "fontsize": 10,
        "fontweight": "bold"
    },
    pctdistance=0.7,
    labeldistance=1.1,
    counterclock=False,
    rotatelabels=False
)

plt.title("Daily Study Time")
plt.legend()
plt.show()
```

```python
# | Parameter       | How to write                       | Why?                                             |
# | --------------- | ---------------------------------- | ------------------------------------------------- |
# | `x`             | `x=[10,20,30]`                     | To provide values for the pie chart.              |
# | `labels`        | `labels=["A","B","C"]`             | To give names to the slices.                      |
# | `autopct`       | `autopct="%1.1f%%"`                | To display percentages on the slices.             |
# | `startangle`    | `startangle=90`                    | To set the starting angle of the pie chart.       |
# | `explode`       | `explode=[0,0.1,0]`                | To separate and highlight a slice.                |
# | `shadow`        | `shadow=True`                      | To add a shadow effect to the pie chart.          |
# | `colors`        | `colors=["red","blue","green"]`    | To assign different colors to the slices.         |
# | `wedgeprops`    | `wedgeprops={"edgecolor":"black"}` | To customize the border and appearance of slices. |
# | `textprops`     | `textprops={"fontsize":12}`        | To customize the label and percentage text.       |
# | `pctdistance`   | `pctdistance=0.7`                  | To control the position of percentage text.       |
# | `labeldistance` | `labeldistance=1.1`                | To control the position of labels.                |
# | `counterclock`  | `counterclock=False`               | To control the direction of the slices.           |
# | `rotatelabels`  | `rotatelabels=True`                | To rotate the labels.                             |
# | `normalize`     | `normalize=True`                   | To normalize the values into a complete pie.      |
# | `radius`        | `radius=1.2`                       | To control the size of the pie chart.             |
# | `center`        | `center=(1,1)`                     | To change the position of the pie chart.          |
```
