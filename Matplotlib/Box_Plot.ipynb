```python
import matplotlib.pyplot as plt 
python = [45, 50, 55, 60, 62, 65, 68, 70, 72, 75, 78, 80, 85, 90, 95]
dsa = [40, 45, 50, 52, 55, 58, 60, 62, 65, 68, 70, 72, 75, 78, 82]
dbms = [35, 45, 48, 50, 55, 58, 60, 65, 68, 70, 72, 75, 80, 85, 100]
plt.boxplot([python,dsa,dbms],labels=["python","dsa","dbms"],
           showmeans=True,
           showfliers=True)
plt.grid(axis="y",color="black",alpha=0.3)
plt.title("Job-level data analysis")
plt.xlabel("subjects")
plt.ylabel("marks")

plt.show()
```
