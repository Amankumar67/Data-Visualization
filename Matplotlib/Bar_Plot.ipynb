```python
# Monthly Sales 

# Requirement: Create a bar chart with different bar colors, a suitable title, x-axis and y-axis labels, and a grid on the y-axis.

import matplotlib.pyplot as plt
month = ["january","february","march","april","may","june"]
sales = [120,150,180,140,200,220]
c = ["Black","brown","darkorange","green","purple","blue"]
plt.bar(month, sales,color = c)
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales Score")
plt.grid(axis="y",linestyle = "--", alpha = 0.6)
plt.show()
```

# Production Comparison 

```python
# Requirement: Create a bar chart and customize the width, align, edgecolor, and linewidth. Also add a title and axis labels.

import matplotlib.pyplot as plt 
product_names = ["prod1","prod2","prod3","prod4","prod5"]
products = [500, 750, 650, 900, 800]
plt.bar(product_names,products,
       width = 0.8,
       align = "center",
       edgecolor = "black",
       linewidth = 2)
plt.title("Production Comparison")
plt.xlabel("product_names")
plt.ylabel("Products")

plt.show()
```

# Student Wise marks 

```python
# Requirement: Create a horizontal bar chart using plt.barh(). Add labels, a title, grid lines

import matplotlib.pyplot as plt 
subjects = ["Python","Maths","DSA","DBMS","AI"]
marks = [88,76,82,79,91]
plt.figure(figsize =(10,6))
plt.barh(subjects, marks,label = "score",color= "darkblue")
plt.title("Subject-Wise Marks Distribution")
plt.grid(axis = "y",linestyle="dashed",alpha =0.5)
plt.legend()
plt.show()
```

# Department Employees 

```python
# Requirement: Create a bar chart with the departments arranged from lowest to highest value. Use custom colors, width, edgecolor, alpha, title, and axis labels.
# IT = 45, HR = 25, Sales = 60, Finance = 35, Marketing = 40.


import matplotlib.pyplot as plt

departments = ["HR", "Finance", "Marketing", "IT", "Sales"]
employees = [25, 35, 40, 45, 60]
c = ["skyblue", "orange", "green", "steelblue", "purple"]
plt.bar(
    departments,
    employees,
    color=c,
    width=0.6,
    edgecolor="black",
    alpha=0.8
)

plt.title("Employees by Department")
plt.xlabel("Departments")
plt.ylabel("Number of Employees")

plt.show()

```

```python
import matplotlib.pyplot as plt
import numpy as np
students = ["A", "B", "C", "D", "E"]
python_marks = [85, 78, 92, 75, 88]
java_marks = [72, 80, 85, 70, 82]

x = np.arange(len(students))
width = 0.35

plt.bar(
    x - width/2,
    python_marks,
    width,
    color="steelblue",
    label="Python"
)
plt.bar(
    x + width/2,
    java_marks,
    width,
    color="orange",
    label="Java"
)
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Python --vs-- Java Marks")
plt.xticks(x, students)
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.show()
```

```python
import matplotlib.pyplot as plt 
x = ["python", "java","c", "C++", "javascript"]
y = [80,40,45,55,65]
c = ["red","blue","green","magenta","purple"]
label = ["python","java","c","C++","javascript"]
plt.bar(x,y, color = c,edgecolor = "black",label = label,linewidth = 2,alpha = 0.9)
plt.title("Programming Language Usage",color = "blue")
plt.xlabel("Languages",fontsize=15,color = "blue")
plt.ylabel("No. of Students",fontsize = 15, color = "blue")
plt.ylim(0,100)
plt.yticks(rotation = 30,color= "red")
plt.xticks(rotation = 30,color = "green")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x = ["python","c","c++","java"]
y = [10,20,30,40]
z = [20,30,40,50]
z1 = [17,35,25,45]
z2 = [25,50,55,30]

width = 0.2
p = np.arange(len(x))
p1 = [j+width for j in p]
p2 = [j+width for j in p1]
p3 = [j+width for j in p2]

plt.xlabel("Languages")
plt.ylabel("numbers")
plt.title("Popularity")

plt.bar(p,y,label="popular",width=width,align="center",edgecolor='black',alpha=0.5)
plt.bar(p1,z,label="less-popular",width=width,align="center",edgecolor='black',alpha=0.7)
plt.bar(p2,z1,label="normal-popular",width=0.2,align="center",edgecolor='black',alpha=0.9)
plt.bar(p3,z2,label="extract-popular",width=0.2,align="center",edgecolor='black',alpha=0.8)

plt.xticks(p + 2 * width, x)  # p + 1.5 for three 
plt.legend()
plt.show()
```
