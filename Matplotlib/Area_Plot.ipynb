# Line chart + filled area = Area Plot

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]

python = [20, 30, 40, 50]
sql = [10, 20, 30, 40]

plt.stackplot(
    months,
    python,
    sql,
    labels=["Python", "SQL"]
)

plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May"]
python = [20, 30, 40, 50, 60]
sql = [10, 20, 25, 35, 45]
plt.stackplot(months,python,sql,labels=["python","sql"])
plt.legend(loc="upper left")
plt.grid(axis="y",color="blue",alpha=0.5)
plt.show()
```

```python
import matplotlib.pyplot as plt
week = ["sunday","monday","tuesdasy","wednesday","thursday"]
python = [20, 30, 40, 50, 60]
sql = [15, 25, 30, 40, 50]
java = [10, 15, 20, 25, 30]
plt.stackplot(week,python,sql,java)
plt.legend(labels=["python","sql","java"],loc="upper left")
plt.grid(axis="y", color="black",alpha=0.5,linestyle="--")
plt.show()
```

```python
import matplotlib.pyplot as plt
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

python = [100, 120, 150, 180, 200, 240]
sql = [80, 90, 110, 130, 150, 170]
ml = [40, 60, 80, 100, 130, 160]
plt.stackplot(months, python, sql , ml,baseline='zero',alpha=0.7,hatch="//")
plt.plot(months,python,marker=">",markerfacecolor="black",markeredgecolor="green")
plt.legend(labels=["python","sql","ml"],loc="upper left")
plt.grid(axis="y",linestyle="--",color="blue")
plt.show()
```

```python
import matplotlib.pyplot as plt 
data = {
    "months": ["Jan", "Feb", "Mar"],
    "python": [20, 30, 40],
    "sql": [10, 20, 30]
}

plt.stackplot(
    "months",
    "python",
    "sql",
    data=data
)
```

```python
import matplotlib.pyplot as plt 
days = ["mon","tue","wed","thu","fri","sat","sun"]
Laptop = [20,25,30,28,35,40,45]
Mobile = [35,40,38,42,45,50,55]
Tablet = [15,18,20,22,25,27,30]
plt.stackplot(days,Laptop,Mobile,Tablet)
plt.plot(Laptop,color="black",marker="o")
plt.grid(axis="y",color="black",alpha = 0.6)
plt.title("Market Sales")
plt.xlabel("days")
plt.ylabel("sales")
plt.legend(labels=["Laptop","Mobile","Tablet"],loc="upper left")
plt.show()
```

```python
import matplotlib.pyplot as plt 
month = ["jan","feb","mar","apr","may","june"]
Google = [120,150,180,200,220,250]
Instagram = [80,100,120,140,160,190]
YouTube = [60,75,90,110,130,150]
plt.stackplot(month,Google,Instagram,YouTube,baseline="wiggle",labels=["Google","Instagram","YouTube"])
plt.title("Social Media")
plt.xlabel("Months")
plt.ylabel("Daily Working Score")
plt.grid(axis="y",color="black",alpha=0.6)                  # baseline = "wiggle"   for using to take y axis value from - to +... 
plt.legend(loc="upper left")
plt.show()
```

```python
import matplotlib.pyplot as plt 
week = ['mon','tue','wed','thu','fri','sat']
Electronics = [100,120,140,160,180,200]
Clothing = [80,90,110,130,150,170]
Grocery = [60,75,85,100,120,140]
Beauty = [40,50,65,70,85,100]
c = ["purple","green","Brown","Pink"]
plt.stackplot(week,Electronics,Clothing,Grocery,Beauty,colors=c)
plt.scatter(week,Electronics,color="red",linestyle="--",linewidth=4)
plt.title("E-Commerce Detail")
plt.xlabel("week")
plt.ylabel("Weekly Sales Score")
plt.grid(axis="y",color="black",alpha=0.7)
plt.legend(labels=["Electronics","Clothing","Grocery",'Beauty'],loc="upper left")
plt.show()
```

```python
import matplotlib.pyplot as plt 
numbers = [1,2,3,4,5,6]
HR = [20,22,25,23,28,30]
IT = [50,55,60,58,65,70]
Marketing = [30,35,32,40,45,50]
Operations = [40,42,45,48,50,55]

plt.stackplot(numbers,HR,IT,Marketing,Operations)
plt.title("Company Details")
plt.xlabel("No.of Professions")
plt.ylabel("Staff Values in No.")
plt.legend(labels=["HR","IT","Marketing","Operations"],loc="upper left")
plt.grid(color="black",alpha=0.)
plt.show()
```
