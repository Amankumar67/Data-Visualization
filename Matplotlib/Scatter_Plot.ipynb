 # Scatter Plot 

```python
# Scatter plot ka use do numerical variables ke relationship/correlation ko show karne ke liye hota hai.
# it is shows in form of dots.
```

```python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5, 6]
marks = [45, 52, 60, 70, 82, 90]

plt.scatter(hours, marks)

plt.show()
```

```python
# color = is used for chaging the colors , also add all different colors 
# s = is used for measuring the size of a dot . 
# alpha = is used for managing the transparency 

import matplotlib.pyplot as plt 
Age = [10, 12, 14, 16, 18]
Height = [130, 140, 150, 165, 175]
plt.scatter(Age,Height,
            color="blue",
            s = 100,
            alpha = 0.7
            )
plt.show()
```

```python
# marker = is used for changing the dot .

import matplotlib.pyplot as plt 
Age = [10, 12, 14, 16, 18]
Height = [130, 140, 150, 165, 175]
plt.scatter(Age,Height,
            color="blue",
            s = 100,
            alpha = 0.7,
            marker = ">"
            )
plt.show()
```

```python
# | Marker | Shape      |
# | ------ | ---------- |
# | `"o"`  | ● Circle   |
# | `"s"`  | ■ Square   |
# | `"^"`  | ▲ Triangle |
# | `"*"`  | ★ Star     |
# | `"+"`  | + Plus     |
# | `"x"`  | × X        |
# | `"D"`  | ◆ Diamond  |
```

```python
# c = is used for sharing different color according to categories 

import matplotlib.pyplot as plt 
Age = [10, 12, 14, 16, 18]
Height = [130, 140, 150, 165, 175]
plt.scatter(Age, Height,  c = Age)            
plt.colorbar()  # is used for measuring color 
plt.show()
```

```python
# c = is used for sharing different color according to categories 

import matplotlib.pyplot as plt 
Age = [10, 12, 14, 16, 18]
Height = [130, 140, 150, 165, 175]
plt.scatter(Age, Height,  c = Height,  cmap="cool")            
plt.colorbar()  # is used for measuring color 
plt.show()
```

```python
#  common colormaps()
# "viridis"
# "plasma"
# "inferno"
# "cool"
# "hot"
```

```python
# c = is used for sharing different color according to categories 
# vmin = color scale's minimum value 
# vmax = color sclae's maximum value 
# plotnonfinite = it gives permission to plot the infinte values .


import matplotlib.pyplot as plt 
Age = [10, 12, 14, 16, 18]
Height = [130, 140, 150, 165, 175]
plt.scatter(Age, Height,  c = Height,  cmap="cool",vmin=150,vmax=300,   plotnonfinite=True)            
plt.colorbar()  # is used for measuring color 
plt.show()
```

```python
# label = is used for given name to plotting. is show by using plt.legend()
# linewidths = points ke border ki thickness ke liye use hota hai ...

import matplotlib.pyplot as plt 
Age = [10, 12, 14, 16, 18]
Height = [130, 140, 150, 165, 175]
plt.scatter(Age,Height,
            color="blue",
            s = 100,
            alpha = 0.7,
            marker = ">",
            label = "height measurement",
            linewidths = 2
            )
plt.legend()
plt.show()
```

```python
# norm = is used for c map for normalize/scale

import matplotlib.pyplot as plt 
from matplotlib.colors import Normalize  # load normalize 
Age = [10, 12, 14, 16, 18]
Height = [130, 140, 150, 165, 175]
norm = Normalize(vmin=130, vmax=175)   # values of normalize 
plt.scatter(Age, Height, 
            c = Height, 
            cmap="cool",
           norm = norm ,    # fit the values of normalize 
           picker = True,     #picker ka use points ko interactive/selectable banane ke liye hota hai
           )
plt.colorbar(label="Height")     # for color parameter 
plt.grid(zorder=3)     #zorder → layer/order of the plot element.
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [0, 2, 3, 4, 5]
y = [10, 15, 20, 25, 30]

plt.scatter(x, y, s=80,color="pink")
plt.plot(x, y, marker="o",markerfacecolor="black",markeredgecolor="green")

plt.title("Scatter + Line")
plt.ylim(0,30)
plt.xlim(0,5)
plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ["Python", "DSA", "DBMS", "Java"]
marks = [90, 75, 80, 65]

plt.bar(subjects, marks, alpha=0.5)
plt.scatter(subjects, marks, s=100)
plt.plot(subjects,marks)

plt.title("Scatter + Bar + Line")
plt.show()
```

```python
import matplotlib.pyplot as plt

data = [10, 12, 14, 15, 15, 16, 18, 20, 22]

plt.hist(data, alpha=0.5, bins=5,color="red")
plt.scatter(data, [1] * len(data), s=80,color="black")

plt.title("Histogram + Scatter")
plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ["Python", "DSA", "DBMS", "Java"]
marks = [90, 75, 80, 65]

plt.barh(subjects, marks, alpha=0.5)
plt.scatter(marks, subjects, s=100)
plt.plot(marks,subjects,color="black")

plt.title("Scatter + Horizontal Bar")
plt.show()
```

```python
# | Parameter       | How to write          | Why?                                                              |
# | --------------- | --------------------- | ----------------------------------------------------------------- |
# | `x`             | `x=[1,2,3]`           | To provide values for the X-axis.                                 |
# | `y`             | `y=[10,20,30]`        | To provide values for the Y-axis.                                 |
# | `s`             | `s=100`               | To control the size of the points.                                |
# | `c`             | `c=[10,20,30]`        | To assign colors based on data values.                            |
# | `marker`        | `marker="o"`          | To change the shape of the points.                                |
# | `cmap`          | `cmap="viridis"`      | To choose the color map for data-based colors.                    |
# | `norm`          | `norm=Normalize(...)` | To normalize data values before mapping them to colors.           |
# | `vmin`          | `vmin=10`             | To set the minimum value of the color scale.                      |
# | `vmax`          | `vmax=100`            | To set the maximum value of the color scale.                      |
# | `alpha`         | `alpha=0.5`           | To control the transparency of points.                            |
# | `linewidths`    | `linewidths=2`        | To control the thickness of point borders.                        |
# | `edgecolors`    | `edgecolors="black"`  | To set the border color of points.                                |
# | `color`         | `color="red"`         | To set a fixed color for the points.                              |
# | `label`         | `label="Students"`    | To give a name to the scatter series for the legend.              |
# | `plotnonfinite` | `plotnonfinite=True`  | To allow non-finite values such as NaN or infinity to be plotted. |
# | `data`          | `data=df`             | To use values directly from a DataFrame or other data object.     |
# | `zorder`        | `zorder=3`            | To control which plot elements appear in front or behind.         |
# | `picker`        | `picker=True`         | To make points selectable/clickable in interactive plots.         |




# | Function     | How to write                 | Why?                                         |
# | ------------ | ---------------------------- | -------------------------------------------- |
# | `colorbar()` | `plt.colorbar()`             | To show the value represented by the colors. |
# | `legend()`   | `plt.legend()`               | To display labels for plotted series.        |
# | `grid()`     | `plt.grid()`                 | To add grid lines for easier reading.        |
# | `xlabel()`   | `plt.xlabel("Age")`          | To name the X-axis.                          |
# | `ylabel()`   | `plt.ylabel("Height")`       | To name the Y-axis.                          |
# | `title()`    | `plt.title("Age vs Height")` | To give the chart a title.                   |

```
